```python
import copy
import random
from pathlib import Path

import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F

from sklearn.metrics import mean_squared_error, r2_score
from sklearn.preprocessing import StandardScaler
from torch.utils.data import DataLoader, TensorDataset
from IPython.display import display, Image


FEATURES = ["RH", "WD", "SR", "BP", "AT"]
TARGET = "PM2.5"
SEQUENCE_LENGTH = 24
EPOCHS = 50
BATCH_SIZE = 32
PATIENCE = 8

OUTPUT_DIR = Path("/content/pyraformer_results")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)


def set_seed(seed=42):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)

    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def load_data(file_path):
    df = pd.read_excel(file_path)
    df.columns = [str(c).strip() for c in df.columns]

    required = [TARGET] + FEATURES
    missing = [c for c in required if c not in df.columns]

    if missing:
        raise ValueError(
            f"Missing columns: {missing}\n"
            f"Available columns: {list(df.columns)}"
        )

    df = df[required].copy()

    for c in required:
        df[c] = pd.to_numeric(df[c], errors="coerce")

    df[FEATURES] = df[FEATURES].interpolate(
        method="linear",
        limit_direction="both"
    )

    df = df.dropna(subset=required).reset_index(drop=True)

    if len(df) < 100:
        raise ValueError("Not enough valid rows for model training.")

    return df


def create_pyramid_mask(level_lengths, radius=2):
    total = sum(level_lengths)

    mask = torch.ones(total, total, dtype=torch.bool)

    offsets = []
    offset = 0

    for length in level_lengths:
        offsets.append(offset)
        offset += length

    for level, length in enumerate(level_lengths):
        start = offsets[level]

        for i in range(length):
            left = max(0, i - radius)
            right = min(length, i + radius + 1)

            mask[
                start + i,
                start + left:start + right
            ] = False

    for level in range(len(level_lengths) - 1):
        fine_len = level_lengths[level]
        coarse_len = level_lengths[level + 1]

        fine_offset = offsets[level]
        coarse_offset = offsets[level + 1]

        for i in range(fine_len):
            parent = i // 2

            if parent < coarse_len:
                fine_node = fine_offset + i
                coarse_node = coarse_offset + parent

                mask[fine_node, coarse_node] = False
                mask[coarse_node, fine_node] = False

    return mask


class PyraformerPM25(nn.Module):

    def __init__(
        self,
        input_features=5,
        sequence_length=24,
        d_model=64,
        n_heads=4,
        num_layers=2,
        dropout=0.1
    ):
        super().__init__()

        if sequence_length % 4 != 0:
            raise ValueError(
                "Sequence length must be divisible by 4."
            )

        if d_model % n_heads != 0:
            raise ValueError(
                "d_model must be divisible by n_heads."
            )

        self.sequence_length = sequence_length

        self.input_projection = nn.Linear(
            input_features, d_model
        )

        self.position_embedding = nn.Parameter(
            torch.randn(1, sequence_length, d_model) * 0.02
        )

        self.scale_embedding = nn.Parameter(
            torch.randn(3, d_model) * 0.02
        )

        self.level_lengths = [
            sequence_length,
            sequence_length // 2,
            sequence_length // 4
        ]

        self.register_buffer(
            "attention_mask",
            create_pyramid_mask(self.level_lengths)
        )

        encoder_layer = nn.TransformerEncoderLayer(
            d_model=d_model,
            nhead=n_heads,
            dim_feedforward=d_model * 4,
            dropout=dropout,
            activation="gelu",
            batch_first=True,
            norm_first=True
        )

        self.encoder = nn.TransformerEncoder(
            encoder_layer,
            num_layers=num_layers
        )

        self.final_norm = nn.LayerNorm(d_model * 3)

        self.regression_head = nn.Sequential(
            nn.Linear(d_model * 3, d_model),
            nn.GELU(),
            nn.Dropout(dropout),
            nn.Linear(d_model, 1)
        )

    def forward(self, x):
        current = (
            self.input_projection(x)
            + self.position_embedding
        )

        levels = []

        for level in range(3):
            if level > 0:
                current = F.avg_pool1d(
                    current.transpose(1, 2),
                    kernel_size=2,
                    stride=2
                ).transpose(1, 2)

            current = (
                current
                + self.scale_embedding[level].view(1, 1, -1)
            )

            levels.append(current)

        pyramid = torch.cat(levels, dim=1)

        encoded = self.encoder(
            pyramid,
            mask=self.attention_mask
        )

        summaries = []
        offset = 0

        for length in self.level_lengths:
            summaries.append(
                encoded[:, offset + length - 1, :]
            )
            offset += length

        representation = torch.cat(summaries, dim=-1)
        representation = self.final_norm(representation)

        return self.regression_head(
            representation
        ).squeeze(-1)


def create_sequences(x, y, start, end, sequence_length):
    sequences = []
    targets = []

    for index in range(
        max(sequence_length - 1, start), end
    ):
        sequences.append(
            x[index - sequence_length + 1:index + 1]
        )
        targets.append(y[index])

    return (
        np.asarray(sequences, dtype=np.float32),
        np.asarray(targets, dtype=np.float32)
    )


def calculate_metrics(actual, predicted):
    mse = mean_squared_error(actual, predicted)
    rmse = np.sqrt(mse)

    denominator = np.maximum(np.abs(actual), 1e-8)
    mape = np.mean(
        np.abs((actual - predicted) / denominator)
    ) * 100

    r2 = r2_score(actual, predicted)

    return {
        "MSE": float(mse),
        "MAPE (%)": float(mape),
        "RMSE": float(rmse),
        "R2": float(r2)
    }


set_seed(42)

df = load_data(data_file)

print("Dataset shape:", df.shape)
print("Input variables:", FEATURES)
print("Target variable:", TARGET)
print("Missing values after preprocessing:")
print(df.isnull().sum())

x_raw = df[FEATURES].to_numpy(dtype=np.float32)
y_raw = df[TARGET].to_numpy(dtype=np.float32).reshape(-1, 1)

n_rows = len(df)

train_end = int(n_rows * 0.72)
test_start = int(n_rows * 0.80)

x_scaler = StandardScaler()
y_scaler = StandardScaler()

x_scaler.fit(x_raw[:train_end])
y_scaler.fit(y_raw[:train_end])

x_scaled = x_scaler.transform(x_raw).astype(np.float32)
y_scaled = y_scaler.transform(y_raw).astype(np.float32).ravel()

x_train, y_train = create_sequences(
    x_scaled, y_scaled,
    0, train_end, SEQUENCE_LENGTH
)

x_val, y_val = create_sequences(
    x_scaled, y_scaled,
    train_end, test_start, SEQUENCE_LENGTH
)

x_test, y_test = create_sequences(
    x_scaled, y_scaled,
    test_start, n_rows, SEQUENCE_LENGTH
)

train_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(x_train),
        torch.from_numpy(y_train)
    ),
    batch_size=BATCH_SIZE,
    shuffle=True
)

val_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(x_val),
        torch.from_numpy(y_val)
    ),
    batch_size=BATCH_SIZE,
    shuffle=False
)

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nDevice:", device)
print("Training samples:", len(x_train))
print("Validation samples:", len(x_val))
print("Testing samples:", len(x_test))

model = PyraformerPM25(
    input_features=len(FEATURES),
    sequence_length=SEQUENCE_LENGTH
).to(device)

criterion = nn.MSELoss()

optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=0.001,
    weight_decay=0.0001
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=3
)

best_val_loss = float("inf")
best_state = copy.deepcopy(model.state_dict())
wait = 0

history = {
    "train_loss": [],
    "val_loss": []
}

print("\nStarting model training...\n")

for epoch in range(EPOCHS):
    model.train()
    train_losses = []

    for batch_x, batch_y in train_loader:
        batch_x = batch_x.to(device)
        batch_y = batch_y.to(device)

        optimizer.zero_grad(set_to_none=True)

        predictions = model(batch_x)
        loss = criterion(predictions, batch_y)

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(), max_norm=1.0
        )

        optimizer.step()
        train_losses.append(loss.item())

    model.eval()
    val_losses = []

    with torch.no_grad():
        for batch_x, batch_y in val_loader:
            batch_x = batch_x.to(device)
            batch_y = batch_y.to(device)

            predictions = model(batch_x)
            loss = criterion(predictions, batch_y)

            val_losses.append(loss.item())

    train_loss = float(np.mean(train_losses))
    val_loss = float(np.mean(val_losses))

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)

    scheduler.step(val_loss)

    print(
        f"Epoch {epoch + 1:02d}/{EPOCHS} | "
        f"Training Loss: {train_loss:.5f} | "
        f"Validation Loss: {val_loss:.5f}"
    )

    if val_loss < best_val_loss:
        best_val_loss = val_loss
        best_state = copy.deepcopy(model.state_dict())
        wait = 0
    else:
        wait += 1

        if wait >= PATIENCE:
            print("Early stopping activated.")
            break

model.load_state_dict(best_state)
model.eval()

test_loader = DataLoader(
    TensorDataset(
        torch.from_numpy(x_test),
        torch.from_numpy(y_test)
    ),
    batch_size=BATCH_SIZE,
    shuffle=False
)

predictions_scaled = []
actual_scaled = []

with torch.no_grad():
    for batch_x, batch_y in test_loader:
        batch_x = batch_x.to(device)

        predictions = model(batch_x).cpu().numpy()

        predictions_scaled.extend(predictions.tolist())
        actual_scaled.extend(batch_y.numpy().tolist())

y_pred = y_scaler.inverse_transform(
    np.array(predictions_scaled).reshape(-1, 1)
).ravel()

y_true = y_scaler.inverse_transform(
    np.array(actual_scaled).reshape(-1, 1)
).ravel()

metrics = calculate_metrics(y_true, y_pred)

metrics_df = pd.DataFrame(
    list(metrics.items()),
    columns=["Metric", "Value"]
)

predictions_df = pd.DataFrame({
    "Dataset_Row_Index": np.arange(test_start, n_rows),
    "Actual_PM2.5": y_true,
    "Predicted_PM2.5": y_pred,
    "Absolute_Error": np.abs(y_true - y_pred)
})

metrics_df.to_csv(
    OUTPUT_DIR / "evaluation_metrics.csv",
    index=False
)

predictions_df.to_csv(
    OUTPUT_DIR / "test_predictions.csv",
    index=False
)

torch.save(
    {
        "model_state_dict": model.state_dict(),
        "sequence_length": SEQUENCE_LENGTH,
        "feature_columns": FEATURES,
        "target_column": TARGET,
        "d_model": 64,
        "n_heads": 4,
        "num_layers": 2
    },
    OUTPUT_DIR / "pyraformer_pm25_best.pth"
)

joblib.dump(
    x_scaler,
    OUTPUT_DIR / "feature_scaler.joblib"
)

joblib.dump(
    y_scaler,
    OUTPUT_DIR / "target_scaler.joblib"
)

plt.figure(figsize=(9, 5))
plt.plot(history["train_loss"], label="Training Loss")
plt.plot(history["val_loss"], label="Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Scaled MSE Loss")
plt.title("Training and Validation Loss")
plt.legend()
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "training_loss.png", dpi=150)
plt.close()

plt.figure(figsize=(12, 5))
plt.plot(y_true, label="Actual PM2.5")
plt.plot(y_pred, label="Predicted PM2.5")
plt.xlabel("Test Sample")
plt.ylabel("PM2.5")
plt.title("Actual vs Predicted PM2.5")
plt.legend()
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "actual_vs_predicted.png", dpi=150)
plt.close()

plt.figure(figsize=(6, 6))
plt.scatter(y_true, y_pred, alpha=0.65)

lower = min(np.min(y_true), np.min(y_pred))
upper = max(np.max(y_true), np.max(y_pred))

plt.plot(
    [lower, upper],
    [lower, upper],
    linestyle="--"
)

plt.xlabel("Actual PM2.5")
plt.ylabel("Predicted PM2.5")
plt.title("Actual vs Predicted Scatter Plot")
plt.tight_layout()
plt.savefig(OUTPUT_DIR / "prediction_scatter.png", dpi=150)
plt.close()

print("\n" + "=" * 50)
print("FINAL EVALUATION RESULTS")
print("=" * 50)

display(metrics_df)

print("\nSaved result files:")
for file_path in sorted(OUTPUT_DIR.iterdir()):
    print(file_path.name)

for name in [
    "training_loss.png",
    "actual_vs_predicted.png",
    "prediction_scatter.png"
]:
    print("\n" + name)
    display(Image(filename=str(OUTPUT_DIR / name)))
```
